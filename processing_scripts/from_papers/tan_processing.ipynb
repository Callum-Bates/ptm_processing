```python
import pandas as pd

# calculated q values from PEP scores - obsidian notes for link to formula paper
# decided to use q value where applicable as less conservative than PEP while still be rigorous 
df = df.sort_values("PEP").reset_index(drop=True)
df["fdr"] = df["PEP"].expanding().mean()
# q-value: smallest FDR at which this item would be accepted
df["qvalue"] = df["fdr"][::-1].cummin()[::-1]



sub = df[df["PEP"] > 0.01]
print(len(sub))
```
