## Data from DN
reports train and test data separate
positives and negatives are separated on 2 different sheets in each file
so far i have just accessed positives 
these also contain mouse data so need to run organism checks 

```python
import pandas as pd

def clean_DN_data(file_name, sheet): 
    df = pd.read_excel(f"/home/callum/sno_data/raw/dn_{file_name}.xlsx", header = 2, sheet_name= sheet)
    print(df.shape)
    #df = df[["UniProt", "Position"]]
    suffixes = ["", ".1", ".2"]
    for i in suffixes:
        df.rename(columns={f"UniProt{i}": f"id{i}", f"Position{i}": f"position{i}"}, inplace=True)
    #print(df.columns)

    dfs = []

    for i in suffixes:
        dfa = df[[f"id{i}", f"position{i}"]].copy()
        dfa.columns = ["id", "position"]
        dfs.append(dfa)

    df_melted = pd.concat(dfs, ignore_index=True)
    df_melted = df_melted.dropna(subset=["id"])
    df_melted["position"] = df_melted["position"].astype(int)
    
    return df_melted

```

```python
# use function for training and test data
train = clean_DN_data("training", "S-Nitrosylation positive")
test = clean_DN_data("test", "S-Nitrosylation positive")

print(train.shape, test.shape)

train.to_csv("/home/callum/sno_data/processed/dn/dn_train_processed.csv", index = False)
test.to_csv("/home/callum/sno_data/processed/dn/dn_test_processed.csv", index = False)
```
